```python
import urllib
import pandas as pd
```

# This notebook enriches the df_votes
### We add topics, periods and other interesting stuff so we can build the graphs based on a specific sorting

```python
with urllib.request.urlopen("https://raw.githubusercontent.com/Somon8/social_graphs_25/main/final-project/graphs/df_votes.csv") as response:
    df = pd.read_csv(response)
```

```python
# Add topic
# Add period
# Add other stuff

with urllib.request.urlopen("https://raw.githubusercontent.com/Somon8/social_graphs_25/main/final-project/folketinget_votings_enriched.csv") as response:
    df_votings = pd.read_csv(response)
```

```python
relevant_cols_from_votings = ['Period', 'id_afstemning'] #Topic etc.
#Here we can also add what party the politician belonged to at the time

df_enriched = df.merge(df_votings[relevant_cols_from_votings], how='left', left_on='voting_id', right_on='id_afstemning')

df_enriched.drop(columns = ['id_afstemning'], inplace=True) #We just needed it for the join
```
