```python
import urllib
import pandas as pd
```

# This notebook merges votes, voting_sessions and actor information

```python
with urllib.request.urlopen("https://raw.githubusercontent.com/Somon8/social_graphs_25/main/final-project/voting-data/df_votes_all_periods.csv") as response:
    df_votes = pd.read_csv(response)

with urllib.request.urlopen("https://raw.githubusercontent.com/Somon8/social_graphs_25/main/final-project/voting-data/voting_sessions_enriched.csv") as response:
    df_voting_sessions = pd.read_csv(response)

with urllib.request.urlopen("https://raw.githubusercontent.com/Somon8/social_graphs_25/main/final-project/actor-data/actor_df.csv") as response:
    df_actors = pd.read_csv(response)
```

```python
df_votes.head(2)
```

```python
relevant_topics_for_joining_on_votes = ["afstemning_id", "afstemning_nummer", "afstemning_vedtaget", "Period", "primary_topic", "all_topics"]
df_voting_sessions.head(2)
```

```python
df_actors.head(2)
```

```python
votes_with_actors = df_votes.merge(df_actors, how = "left",on = "aktørid") #THIS IS GOING TO NEED AN UPDATE WITH THE CHANGING PARTIES BUT WHATEVER
votes_with_actors.head(2)
```

```python
df_voting_sessions[relevant_topics_for_joining_on_votes]
```

```python
votes_with_actors_enriched = votes_with_actors.merge(df_voting_sessions[relevant_topics_for_joining_on_votes], how = "left", on = "afstemning_id")
votes_with_actors_enriched.head(2)
```

```python
votes_with_actors_enriched.to_csv("./voting-data/all_votes_enriched.csv", index = False)
```
